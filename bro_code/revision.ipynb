```python
#compund interest calculator
princple = float(input("Enter the ammount"))
rate = float(input("Enter the rate"))/100
time = float(input("enter the time."))
while princple == "":
    print("Please enter the principle amount: ")
    princple = float(input("Enter the ammount"))
while rate == "":
    print("please enter the rate: ")
    rate = float(input("Enter the rate"))/100
while time =="":
    print("Please enter the time: ")
    time = float(input("enter the time."))

final_amount = princple*((1+rate)**time)
print(f"{final_amount}")
```
