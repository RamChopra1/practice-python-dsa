```python
# *args ==> allow you to pass multiple non-key arguments
# **kwarge ==> allows you pass multiple keyword-arguments
#  *unpacking operator 
#    1 . positional ,2. DEFAULT , 3. keyword , 4. arbitary

def add(*args):
    total = 0 
    for arg in args:         # main thing is "*" before parameter name 
        total += arg
    return total
print(add(1,2,3))
```
